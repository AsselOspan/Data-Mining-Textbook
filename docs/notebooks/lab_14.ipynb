## Есеп және деректермен қамтылуы

Сондықтан төменде алты жазбадан тұратын жаңа шағын оқу жиыны толық жасалады; оны бастапқы файлдың сол жиіліктері бар қысқартылған бөлігі деп санауға болмайды. Python, pandas және openpyxl қажет. json, pathlib және xml.etree.ElementTree стандартты кітапханаға кіреді. Барлық дереккөз кодпен work_results ішінде жасалады, курстың бастапқы файлдары өзгертілмейді. Төмендегі XML — DTD мен сыртқы мәндері жоқ, өзіміздің шектелген сенімді құжатымыз. Оны желідегі кез келген XML-мен алмастырмаңыз: бөгде XML қауіпсіздігі бөлек талдау мен шектеулерді талап етеді.

## Бастапқы талданған мысал

Бұл мысал тапсырмадан өзгеше бір жазбаны сериализациялау мен оқуды ғана көрсетеді. tags өрісі тізім болып қалады. dump ашық файлға жазады, load файлдан оқиды. Сәтті оқығаннан кейін синтаксисті ғана емес, күтілетін құрылымды да тексеру қажет.



```python
from pathlib import Path
import json
import xml.etree.ElementTree as ET
import pandas as pd

out = Path('work_results')
out.mkdir(exist_ok=True)
demo = {'id': 1, 'title': 'Оқу жазбасы', 'tags': ['a', 'b']}
with (out / 'lab14_demo.json').open('w', encoding='utf-8') as f:
    json.dump(demo, f, ensure_ascii=False, indent=2,
              allow_nan=False)
with (out / 'lab14_demo.json').open(encoding='utf-8') as f:
    reread = json.load(f)
assert reread == demo
print(type(reread).__name__, reread['title'])
```


Нәтиже — dict және «Оқу жазбасы» тақырыбы. Бұл DataFrame емес, бірақ жазбаны кейін кестеге орналастыруға болады. JSON жоғарғы деңгейі массив болса, парсер list қайтарар еді. Десериализацияға eval қолданбаңыз: файлда орындауға рұқсат берілген командалар емес, деректер бар.

## Толық жергілікті дереккөздер

Барлық негізгі тапсырмаға арналған JSON алу үшін төмендегі екі блокты орындаңыз. id — бірегей бүтін сан, cuisine — асүйдің оқу санаты, ingredients — әртүрлі мәтіндік ингредиенттердің бос емес тізімі. Жаттығу аударма сөздігіне тәуелді болмауы үшін шартты ағылшын белгілері қолданылады. Жиында нақты рецепт ұсынымдары жоқ.



```python
recipes = [
    {'id': 101, 'cuisine': 'local',
     'ingredients': ['rice', 'carrot', 'onion']},
    {'id': 102, 'cuisine': 'mediterranean',
     'ingredients': ['tomato', 'cheese', 'onion']},
    {'id': 103, 'cuisine': 'local',
     'ingredients': ['rice', 'onion']},
    {'id': 104, 'cuisine': 'asian',
     'ingredients': ['rice', 'ginger', 'carrot']},
    {'id': 105, 'cuisine': 'mediterranean',
     'ingredients': ['tomato', 'pepper', 'cheese']},
    {'id': 106, 'cuisine': 'local',
     'ingredients': ['potato', 'onion', 'carrot']},
]
```

```python
source_json = out / 'lab14_recipes_source.json'
with source_json.open('w', encoding='utf-8') as f:
    json.dump(recipes, f, ensure_ascii=False, indent=2,
              allow_nan=False)
```


XML дереккөзін де берілген деректерден толық жасаңыз. Мәзірде үш тағам бар, әрқайсысында жолдық id және name, шартты бірліктегі бүтін price және граммен берілген weight бар. Бұл рецептердің сыртқы анықтамалығы емес, бөлек кесте: шынайы сәйкестік кілтісіз оны ұқсас атаулар арқылы recipes-пен біріктірмеңіз.



```python
menu_rows = [
    ('m01', 'Салат', 450, 180),
    ('m02', 'Күріш', 600, 250),
    ('m03', 'Көкөністер', 800, 300),
]
menu_root = ET.Element('menu')
for code, name, price, weight in menu_rows:
    node = ET.SubElement(menu_root, 'dish', id=code, name=name)
    ET.SubElement(node, 'price').text = str(price)
    ET.SubElement(node, 'weight').text = str(weight)
ET.ElementTree(menu_root).write(
    out / 'lab14_menu_source.xml', encoding='utf-8',
    xml_declaration=True,
)
```
