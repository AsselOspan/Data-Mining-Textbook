## Деректер және дайындық

Тек pandas, NumPy және жинақтағы data/football.csv қолданылады. Барлық есептеу жергілікті орындалады. Баған атаулары дәл жазылуы тиіс: SprintSpeed, HeadingAccuracy, Composure, Reactions, GKReflexes, GKHandling, ShotPower, Aggression. Wage пен Value — дереккөз бірліктеріндегі көрсеткіштер; валюта мен кезең метадеректер бойынша расталмаған. Name бірегейлігіне кепілдік жоқ; Unnamed: 0 сипаттама ретінде қолданылмайды.


Негізгі есептеулерді бастамай тұрып `.loc[маска, бағандар]` өрнегін оқу ережесін жазыңыз: бірінші бөлігі — маска, екіншісі — бағандар. Әр тапсырма үшін масканы жеке айнымалыға құрып, оның көлемін тексеріңіз. Шектерді шартта көрсетілген жиын бойынша есептеңіз. Жүздікке дейін дөңгелектеуді шығару кезінде орындаңыз. Жоқ нәтижені түсіндірмесіз нөлмен алмастырмаңыз.

## Талданған мысал: шектен қатаң жоғары



```python
demo = pd.DataFrame({
    'Name': ['A', 'B', 'C', 'D'],
    'Age': [18, 24, 24, 30],
    'Rating': [60, 70, 80, 90]
})
threshold = demo['Age'].mean()
mask = demo['Age'] > threshold
print(threshold)
print(demo.loc[mask, 'Name'].tolist())
print(demo.loc[mask, 'Rating'].mean())
```


Орташа жас 24-ке тең; одан қатаң үлкен тек D, орташа Rating 90-ға тең. B және C жазбалары «кіші» тобына да, «үлкен» тобына да кірмеді. Егер сұрақ бүкіл кестені екі топпен қамтуды талап етсе, бір шекараны қатаң емес етіп, оны тікелей көрсету керек. Төмендегі тапсырмаларда шекаралар берілген, оларды өздігінен өзгертуге болмайды.

## Тапсырмалар

### 1. Белгілер, орындар және пішін

`.iloc` және бағандардың айқын тізімі арқылы алғашқы бес жолды және Name, Age, Position бағандарын таңдаңыз. Содан кейін үшінші жолдың индекс белгісін алып, сол жолды `.loc` арқылы таңдаңыз. Age мәндерінің теңдігін көрсетіңіз. Age бағанын бөлек Series және DataFrame ретінде алып, пішіндерін жазыңыз. Age бойынша сұрыптағаннан кейін орын бойынша таңдауды қайталаңыз: орынның енді неге басқа ретке қатысты екенін түсіндіріңіз.

### 2. Wage орташадан жоғары және төмен

Толық football бойынша орташа Wage есептеңіз. Wage > mean және Wage < mean қатаң шарттары бар топтар құрыңыз. Әрқайсысының жазба санын және орташа SprintSpeed табыңыз. Wage орташа мәнге тең жазбалар санын да тексеріңіз; үш санның қосындысы жол санына тең болуы тиіс. Орташа рейтингтер айырмасын жазыңыз. Жауапты Wage жылдамдыққа себептік әсер етеді деп тұжырымдамау керек.

### 3. Максимал Wage және тең мәндер

Wage максимал барлық жолды табыңыз; Name, Club, Position, Wage сақтаңыз. Бірнеше жол болса, кездесетін барлық позицияны және жазба санын келтіріңіз. Бұл тәсілді `idxmax()` арқылы таңдаумен салыстырыңыз: тек бір белгіні қолданғанда ненің жоғалуы мүмкін екенін түсіндіріңіз. Нәтижені индекссіз `work_results/lab02_max_wage.csv` файлына сақтаңыз. Табылған атты бірегей кілт деп жарияламаңыз.

### 4. Penalties туралы сұрақты түзету

Nationality == 'Brazil' жазбалары үшін бақылаулар санын, Penalties минимумын, максимумын, орташа мәнін және медианасын есептеңіз. Есепте «қанша пенальти соқты» деген бастапқы қате түсіндірмені келтіріп, оны бұл жиыннан неліктен есептеу мүмкін еместігін түсіндіріңіз. Кезеңдегі оқиғаларды санауға қажет болатын екі жетіспейтін өрісті ұсыныңыз. Жалған гол санын алу үшін рейтингті қоспаңыз.

### 5. Шектердің екі түрі

HeadingAccuracy > 50 үшін таңдама көлемін және орташа Age табыңыз. Содан кейін толық кесте бойынша 0.9 * max(Composure) және 0.9 * max(Reactions) шектерін есептеңіз. Екі белгісі де өз шегінен қатаң асатын жазбаларды таңдап, минимал Age табыңыз. Ең кіші жастағы жолдарды Name, Age, Composure, Reactions өрістерімен көрсетіңіз. Таңдама бос болса, мағыналы хабарды және шектердің өзін ұсыныңыз. «90%» сөзі рейтинг > 90 шартына емес, максимумның үлесіне қатысты.

### 6. Ең жас және ең егде топтар

Минимал және максимал Age табыңыз. Осы мәндерге теңдік бойынша екі топ құрып, әрқайсысында жазба санын және орташа Reactions есептеңіз. «Үлкендер минус кішілер» айырмасын анықтап, көрсетілуін жүздікке дейін дөңгелектеңіз. Оң мән файлдағы үлкен жастағы топтың орташа рейтингі жоғары екенін, теріс мән керісінше екенін білдіреді. Шеткі жасты салыстыру бір адамның есею әсерін неліктен дәлелдемейтінін көрсетіңіз.

### 7. Қақпашылар және салыстырмалы бағалау

Алдымен Position == 'GK' таңдаңыз. Тек осы таңдама ішінде GKReflexes және GKHandling максимумдарын табыңыз. Әр максимум үшін барлық иесін сақтап, орташа Wage есептеңіз. Содан кейін бірінші орташа мәннің екіншісіне қатынасын табыңыз. Бөлуден бұрын топтардың бос еместігін және бөлімнің нөл еместігін тексеріңіз. Топтар қиылысуы мүмкін: екі максимум иелерінің арасында бірдей бастапқы жолдар бар-жоғын бөлек анықтаңыз. Жасанды тәуелсіздік жасау үшін оларды топтардан алып тастамаңыз.

### 8. Агрессивтілік және соққы күші

Бүкіл кесте үшін Aggression максимал және минимал топтарды табыңыз. Әр топтың орташа ShotPower мәнін және «максимал агрессивтілік / минимал» қатынасын есептеңіз. Екі жазба санын да шығарыңыз. Name, Aggression, ShotPower және жаңа логикалық ExtremeAggression бағаны бар жұмыс кестесін құрыңыз. Меншіктеуді таңдаулар тізбегінсіз, тікелей баған не біртұтас `.loc` арқылы орындаңыз. Бастапқы football кестесін өзгертпеңіз.

### 9. Қорытынды кесте және тексеру

Сандық салыстыру нәтижелерін `work_results/lab02_summary.csv` файлына жинаңыз. Бір жол 2, 4, 5, 6, 7 не 8-тапсырмадағы бір көрсеткішті сипаттауы тиіс. Өрістері: `task`, `metric`, `n_selected`, `value`, `interpretation`. Қатынас үшін n_selected ішінде қатысқан бастапқы жолдардың бірігу көлемін көрсетіңіз, ал түсіндірмеде қосымша екі жазба санын да беріңіз. Көрсеткіш анықталмаса, value жетіспейтін күйде қалып, себебі interpretation ішінде түсіндірілсін.

Екі нәтижені таңдап, басқа түсінікті тәсілмен қайта есептеңіз: мысалы, `.loc[mask, 'Age'].mean()` нәтижесін айқын сақталған Series орташа мәнімен салыстырыңыз. Тексеру бір жиынды салыстыруы және шартты өзгертпеуі тиіс. Соңында барлық есептеуді таза күйден қайта іске қосыңыз.
