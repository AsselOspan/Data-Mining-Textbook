## Дайындық және мысалды өз бетімен талдау



```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

football = pd.read_csv(Path('data') / 'football.csv')
out = Path('work_results') / 'lab08'
out.mkdir(parents=True, exist_ok=True)
assert football.shape == (12897, 42)
```


Кейінгі барлық түрлендіруді жаңа кестелерде орындаңыз. Графикке дейін оның көлемі мен сандық дереккөзін шығарыңыз. notebook ішінде кодты таза күйден ретімен орындаңыз. Скриптпен жұмыс істесеңіз, суретті қалыптастырғаннан кейін show шақырыңыз немесе PNG сақтап, оны қараңыз. Ескертулерді шығаруды бұғаттауды қатені жасыру тәсілі ретінде қолданбаңыз.

Талданған мысал кітап саны туралы жасанды деректерді қолданады; ол кесте мен Axes нысанының байланысын көрсетеді және тапсырмалардан ерекшеленеді.



```python
books = pd.Series([3, 5, 2], index=['A', 'B', 'C'], name='count')
fig, ax = plt.subplots(figsize=(5, 3), layout='constrained')
books.plot.bar(ax=ax, color='slateblue', rot=0)
ax.set(title='Оқу мысалы: сөредегі кітаптар',
       xlabel='Сөре', ylabel='Кітаптар саны')
ax.set_ylim(0, 6)
ax.grid(axis='y', alpha=0.2)
fig.savefig(out / 'example_books.png', dpi=150)
plt.show()
```


Мәндер қосындысы — 10, B сөресіндегі сан ең үлкен. Бұл бақылау кестемен де, кескінмен де расталады. Ол кітаптардың неге осылай бөлінгенін түсіндірмейді. Футболшылармен жұмыс істегенде сипаттау мен түсіндірудің осы айырмасын сақтаңыз. Әр тапсырма үшін алдымен сұрақ түріндегі сөйлем құрып, содан кейін қажетті есептеуді жасаңыз.

Абсолют сан мен үлес бойынша қорытынды неге ерекшеленуі мүмкін екенін түсіндіріңіз.

Есептеуге сүйеніп, көрсетілгендердің ішінде медианасы ең үлкен топты атаңыз және оны орташа мәні ең үлкен топпен салыстырыңыз. Шашыраудың рөлін және медианалардың шағын айырмасы маңыздылықты неге дәлелдемейтінін түсіндіріңіз. Мұрттардан тыс нүктелерді қалдырыңыз: визуализация есебі оларды жоюға рұқсат бермейді.

Оның пайдасын қатар тұрған екі панель нұсқасымен салыстырып, оқуға ыңғайлы нәтижені таңдаңыз.

Оқылуын жақсарту үшін 7-тапсырма кірістірмесін жеке PNG ішінде сақтап, шолу торынан алып тастауға болады. ## Тапсырылатын материалдар және дұрыстық критерийлері

lab08.ipynb немесе lab08.py, plan.csv, сандық CSV файлдарын және кескіндер бумасын тапсырыңыз. Алты негізгі PNG атауы: position, age, position_age, speed_scatter, stamina_box, stamina_range, кеңейтімі .png. Scatter үшін корреляцияларды group, n, pearson_r кестесінде сақтауға болады; бағандар мен жолдарды енгізу ережесі көрсетілсе, бастапқы жұптарды қайта сақтау міндетті емес. Boxplot үшін Position, n, mean, median, q1, q3, minimum, maximum бар stamina_summary.csv сақтаңыз.

Барлық жазуды қазақша жазыңыз, тексеруге көмектесетін жерде бастапқы баған атауларын сақтаңыз. Рейтингтерді пайыз емес, рейтинг деп белгілеңіз. Бірегейлік расталмайынша «жазбалар саны» көрсеткішін бірегей адамдар саны деп атамаңыз. Сақталған кестелер топтарды таңдау мен агрегаттауды толық түсіндіруі тиіс. Оқырман бастапқы кодты қарамай графикті түсінсе, безендіру дұрыс болып есептеледі.

Пайдалы өзіндік тексерулер: жас жиіліктерінің қосындысы көлемге сәйкес; позиция бойынша үлестер қосындысы 1-ге тең; minimum <= q1 <= median <= q3 <= maximum; графиктер мен кестелердегі топ көлемдері бірдей; scatter координаттары бірдей жолдардан алынған. Анықталған корреляция [-1, 1] ішінде болуы тиіс. Бұл тексерулер сәтті түспен безендіру арқылы түзету мүмкін емес дайындық қателерін анықтайды.

Қосымша тапсырма. Acceleration және SprintSpeed үшін scatter мен hexbin салыстырыңыз. Бірдей ось шектерін таңдап, ұяшықтар санының рөлін түсіндіріңіз және қандай тығыздық аймақтары анығырақ көрінгенін көрсетіңіз. Міндетті кескіндерді алмастырмай, қосымша PNG мен шағын абзац қосыңыз.
