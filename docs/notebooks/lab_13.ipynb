Біржақты және екіжақты тексеру нұсқалары ыңғайлы нәтижені таңдау мүмкіндігі ретінде емес, алдын ала қойылған әртүрлі сұрақтар ретінде салыстырылады.

Құралдар: Python, NumPy, pandas, SciPy. Барлық шарттар төменде берілген, football.csv data ішінде орналасқан. Желі қажет емес. Нәтижелер work_results ішіне сақталады. Әр симуляция үшін np.random.default_rng(42) қолданыңыз. Футбол файлын сипаттағанда оның жолдары туралы қорытынды жасаңыз; мақсатты бас жиынтық пен іріктеу механизмі белгісіз.

## Талданған мысал

Қалыпты процестен алынған тәуелсіз таңдамада n=36, орташа мән 102, белгілі sigma=6 болсын. H0: mu=100 гипотезасы H1: mu!=100 баламасына қарсы тексеріледі, alpha=0.05. Стандартты қате 1, сондықтан z=2. Екіжақты p шамамен 0.0455; көрсетілген алғышарттарда H0 жоққа шығарылады. Бағаланған айырма бастапқы бірліктермен +2-ге тең. Оның практикалық мағынасы процестің рұқсат етілетін қатесіне тәуелді.



```python
from pathlib import Path
import numpy as np
import pandas as pd
from scipy.stats import norm, t, ttest_1samp, ttest_ind

out = Path('work_results')
out.mkdir(exist_ok=True)
alpha = 0.05
z = (102 - 100) / (6 / np.sqrt(36))
pvalue = 2 * norm.sf(abs(z))
print('z:', z, 'p:', round(pvalue, 5))
print('H0-ді жоққа шығару:', pvalue < alpha)
```


p-мәнінің дұрыс түсіндірмесі: H0 және берілген модель кезінде екіжақты мағынада одан кем емес экстремал статистика мәнін алу ықтималдығы шамамен 0.0455. Бұл H0 ақиқаттығының ықтималдығы да, дәл біздің қорытындымыз қате болу ықтималдығы да емес. Шешім ережесі алғышарттар орындалғанда қайталама зерттеулердегі рәсім қателерінің жиілігін шектейді.

Бастапқы мәндері бар тапсырмалар үшін мына массивтерді қолданыңыз:



```python
returns = np.array([
    6.1, 9.2, 11.5, 8.6, 12.1,
    3.9, 8.4, 10.1, 9.4, 8.9,
])
service_a = np.array([21., 24., 22., 26., 20., 25., 23., 24.])
service_b = np.array([27., 25., 29., 28., 30., 26., 24., 31.])
football = pd.read_csv(Path('data') / 'football.csv')
assert {'Age', 'Position', 'Club'}.issubset(football.columns)
```


returns мәміле табыстылығының оқу пайыздық мәндерін қамтиды; 10 дегеніміз 10 пайыз, ал орташа мәндер айырмасы пайыздық тармақпен өлшенеді. service_a және service_b — екі жасанды тәуелсіз топтағы операцияны орындау секундтары. Бұл есептерде жуық қалыптылық пен тәуелсіздік оқу шартымен берілген. Қосымша кестелер жүктеу немесе бастапқы суреттерден берілмеген мәндерді қою қажет емес.

H0/H1 жазыңыз және z, сол жақ p, alpha=0.05 кезіндегі шешім мен аймен берілген бағаланған әсерді есептеңіз. H0-ді жоққа шығару да өндірушінің қасақаналығын анықтамайтынын бөлек сөйлеммен түсіндіріңіз. alpha-мен салыстыруда дөңгелектелген p қолданбаңыз.

Деректерді көргеннен кейін маңызды біржақты нұсқаны неліктен «негізгі» деп жариялауға болмайтынын түсіндіріңіз. Топ көлемдерін, орташа мәндерін, A-B айырмасын, статистиканы, p мен шешімді шығарыңыз. Секундпен берілген әсер таңбасын түсіндіріңіз. Содан кейін жұпталған тест қажет болатын дизайнды сипаттаңыз: бір пайдаланушылар екі нұсқаны да орындайды. Тек басқа функция үшін деректерді жасанды жұптарға айналдырмаңыз.

## Не тапсыру керек

Түсіндірмелері бар notebook не скрипт, lab13_tests.csv, lab13_multiple.csv және әр жағдай бойынша қысқа қорытынды тапсырыңыз. | Тексеру | Қажетті инвариант |
|---|---|
| Кез келген p | 0<=p_value<=1 |
| Шешім | reject_h0 мәні p_value<alpha шартына сәйкес |
| Қолмен есептелген t және SciPy | Статистикалар мен p рұқсат шегінде сәйкес |
| Екі топ | effect таңбасы mean_A-mean_B шамасына сәйкес |
| Көптік симуляция | 1000 жол, жоққа шығару сандары 0-ден 20-ға дейін |
| Football | Ықтималдық моделінің оқу сипаты нақты көрсетілген |

lab13_multiple.csv repeat, rejected_005 және rejected_bonferroni — әр зерттеудегі жоққа шығарулар санын қамтуы керек. Кемінде бір жоққа шығаруы бар зерттеулердің қорытынды үлестерін есепте хабарлаңыз. Осы бинарлық индикаторлардың стандартты ауытқуын «p-мәнінің қатесі» деп атамаңыз: бұл басқа кездейсоқ шама.

Жұмыс сапасын жоққа шығарылған H0 саны бойынша бағаламаңыз. Адал түсіндірмесі бар дұрыс жоққа шығармау — толыққанды нәтиже. Деректер алғышарттарға сай болмаса, мұны нақты көрсетіп, қорытындыны шектеңіз; критерий атауын басқаға алмастырудың өзі таңдама ығысуын не тәуелділікті жоймайды.

Қосымша тапсырма: n=10 ішіндегі k=2 сәтті нәтиже үшін p0=0.5-ке қарсы alternative='less' параметрімен scipy.stats.binomtest арқылы дәл биномиалдық тексеру орындаңыз. Қалыпты жуықтаумен салыстырып, күтілетін оқиғалардың аз саны неліктен сақтықты қажет ететінін түсіндіріңіз. Қосымша тапсырма негізгі есептің алдын ала таңдалған тексерулерін өзгертпейді.
