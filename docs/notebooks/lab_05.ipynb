## Құралдар және толық бастапқы кестелер

Python, pandas және pathlib қолданыңыз. Кодты жинақ түбірінен іске қосыңыз. Бұл жұмыста football.csv қажет емес: екі оқиға желісі де толық дербес. Барлық жасалатын файл work_results ішіне жазылады. Алдымен дайындықты орындаңыз:



```python
from pathlib import Path
import pandas as pd

result_dir = Path('work_results')
result_dir.mkdir(exist_ok=True)
items = pd.DataFrame({
    'item_id': [417283, 849734, 132223, 573943, 19475,
                3294095, 382043, 302948, 100132, 312394],
    'vendor': ['Samsung', 'LG', 'Apple', 'Apple', 'LG',
               'Apple', 'Samsung', 'Samsung', 'LG', 'ZTE'],
    'stock_count': [54, 33, 122, 18, 102, 43, 77, 143, 60, 19]
})
```


item_id — тауар моделінің кілті; vendor — өндіруші; stock_count — қордағы бірлік саны. purchase_id — бір сатып алудың әртүрлі жолдары үшін қайталанатын сатып алу нөмірі. line_id оқу нұсқасына журнал жолының бірегей идентификаторы ретінде қосылған, price — шартты ақша бірлігіндегі баға. Сатып алу нөмірі тауар жолының бірегей кілті емес. Төмендегі қорды болжамды есептеу үшін модель бағасы тұрақты деген оқу жорамалы енгізіледі.

Екінші оқиға желісінің толық жасанды кестелері:


rating_id — бағаның бірегей жолы, user_id қайталануы мүмкін. movie_id 999 мәні анықтамалықта әдейі жоқ, ал 104 фильмінің бағалары жоқ. Бұл ерекшеліктерді біле отырып, дайын толық шешімсіз өз нәтижелеріңізді тексере аласыз.

## Талданған мысал: сәйкестік индикаторы



```python
logs = pd.DataFrame({'event': [1, 2], 'code': ['a', 'z']})
lookup = pd.DataFrame({'code': ['a', 'b'], 'label': ['A', 'B']})
example = logs.merge(
    lookup, on='code', how='left',
    validate='many_to_one', indicator=True
)
print(example[['event', 'label', '_merge']])
```


Журналдың екі жолы да қалады. 1 оқиғасында label A-ға және индикатор both мәніне тең, 2 оқиғасында label жоқ және индикатор left_only. Бұл дұрыс left-біріктіру. Жетіспейтін мәннің болуы анықтамалықтың толық еместігін хабарлайды, оқиғаны жоюға рұқсат бермейді. Тапсырмаларды өз items/purchases және ratings/movies кестелеріңізде орындаңыз.

## Тапсырмалар

### 1. Кілттер паспорты

Төрт кестенің бәрі үшін жол мағынасын, кілтті және қайталануы мүмкін өрістерді анықтаңыз. Таңдалған кілттерде жетіспейтін мәндердің жоқтығын және қажет жерде бірегейлігін тексеріңіз. purchases үшін purchase_id неге қайталанатынын және line_id item_id мәнінен қалай ерекшеленетінін түсіндіріңіз. items кестесін purchases кестесіне және movies кестесін ratings кестесіне жалғаудың күтілетін кардиналдылығын алдын ала көрсетіңіз. Нәтижені тек True жиыны емес, шағын кесте түрінде рәсімдеңіз.

### 2. Дүкенді outer арқылы тексеру

purchases сол жақ кестесін items кестесімен item_id бойынша how='outer', indicator=True және дұрыс validate арқылы біріктіріңіз. left_only және right_only жолдарын бөлек алыңыз. Сатуда кездескенімен, қойма туралы жазбасы жоқ модельді және қоймада бар, бірақ сатылмаған модельді анықтаңыз. item_id, side, explanation өрістерімен сәйкестендірілмеген кілттер есебін сақтаңыз. Себеп талданбай тұрып жетіспейтін stock_count мәнін нөлмен алмастырмаңыз.

### 3. Сәйкес модельдер және қосындыны бақылау

purchases пен items үшін inner-біріктіру жасаңыз. Нәтижедегі оқиға жолдарының санын және бірегей модель санын анықтаңыз: бұл шағын журналда сәйкес келуі мүмкін болса да, олар әртүрлі ұғымдар. line_id көбеймегенін тексеріңіз. inner алдында және кейін price қосындысын салыстырып, айырманың қай жолға қатысты екенін түсіндіріңіз. Кейін left орындап, оның бастапқы журналдағы барлық line_id мәнін сақтайтынына көз жеткізіңіз.

### 4. Қорды толық сатудың оқу бағалауы

purchases және items екеуінде де бар модельдер үшін `potential_revenue = stock_count * price` есептеңіз. Мына жорамал енгізіледі: осы оқу журналында модельдің соңғы бақыланған бағасы бүкіл қор сатылғанда тұрақты қалады, жеңілдік пен шығын жоқ. Модельдің бірнеше әртүрлі бағасы болса, алдымен айқын таңдау ережесі қажет болар еді. Дәл осы кіріс деректер үшін баға кестесіндегі item_id бірегейлігін тексеріңіз. potential_revenue максимал барлық модельді және сәйкес модельдердің жалпы ықтимал көлемін табыңыз. Бұл сценарийге бұрын орындалған сатып алулардың нақты түсімін қоспаңыз.

### 5. Бағалар мен фильмдерді біріктіру түрлері

ratings сол жақ, movies оң жақ кесте ретінде left, right, inner және outer орындаңыз. Әр түр үшін жол санын және indicator жиіліктерін көрсетіңіз. Қандай жазбалар сақталатынын және outer неге ratings кестесінен көп жолға ие болуы мүмкін екенін түсіндіріңіз. Сыртқы біріктіруден 104 фильмінің жылын табыңыз, ал 999 үшін атауы мен жылын алу мүмкін еместігін түсіндіріңіз. Оқу анықтамалығында жоқ мәліметті ойдан шығармаңыз.

### 6. Анықтамалықтың бақыланатын қайшылығы

movies көшірмесінен қосымша жолы бар movies_conflict жасаңыз: movie_id = 101, title = 'Қайшы нұсқа', year = 2002. validate параметрінсіз біріктіріп, жол санының өсуін өлшеңіз. Содан кейін validate='many_to_one' арқылы қайталап, pd.errors.MergeError ұстап, түсінікті хабар көрсетіңіз. Кілттің барлық қайшы жазбасын шығарыңыз. Таңдау ережесінсіз movie_id бойынша drop_duplicates неліктен негізделген түзету бермейтінін түсіндіріңіз. Бастапқы movies өзгермесін.

### 7. Жүктеме бөліктерін жасау

`work_results/lab05_parts` каталогын жасап, ratings кестесін екі тізбекті жолдан үш CSV түрінде сақтаңыз: part_01.csv, part_02.csv, part_03.csv. index=False қолданыңыз. Сондай-ақ `['part_01.csv', 'notes.txt', 'part_02.csv.bak', 'part_03.csv']` атаулар тізімін беріп, одан Path(name).suffix арқылы тек .csv кеңейтімді атауларды таңдаңыз. csv ішкі жолын іздеу сақтық көшірмені неліктен қате қосатынын түсіндіріңіз.

### 8. concat арқылы жинау

Жасалған каталогтағы нақты part_*.csv файлдарының сұрыпталған тізімін алыңыз. Әрқайсысын оқып, rating_id, user_id, movie_id, rating дәл схемасын тексеріңіз, source_file атауын қосып, кестелер тізімін жинаңыз да, ignore_index=True арқылы concat бір рет шақырыңыз. Жол санын, rating_id бірегейлігін және source_file жойылып, rating_id бойынша сұрыпталғаннан кейін бастапқы ratings кестесіне теңдігін тексеріңіз. 0.5 бағаларының санын есептеп, оның футбол жиынындағы Penalties дағды рейтингінен айырмасы ретінде оқиғалар санауышы екенін түсіндіріңіз.

### 9. Қорытынды файлдар және сапа есебі

`lab05_unmatched_items.csv`, `lab05_stock_scenario.csv` және `lab05_ratings_joined.csv` файлдарын work_results ішінде сақтаңыз. Соңғысы үшін тексерілген movies анықтамалығын қалпына келтірілген бағаларға left арқылы жалғаңыз; source_file және indicator сақтаңыз. Есепте амалдар алдындағы және кейінгі жол санын, табылған сәйкестендірілмеген кілттерді және байқалған қайшылықты келтіріңіз. Дұрыс left-біріктіру кезінде журнал қосындысының сақталуын тексерудің кемінде бір мысалын көрсетіңіз. Шешімді таза күйден қайта іске қосыңыз.
