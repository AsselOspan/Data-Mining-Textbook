Көшірме бірінші өзгерістен кейін емес, оған дейін жасалуы тиіс. Есепте бастапқы матрицаны тілімдер арқылы өзгеруден қай код жолдары қорғайтынын көрсетіңіз.

## Толық берілген бастапқы массивтер



```python
import numpy as np

first_line = [x * y for x in range(2, 100, 6)
              for y in range(7, 1, -2)]
second_line = [x ** 0.5 for x in range(1000, 1101, 2)]
third_line = [x ** 2 for x in range(51)]
big_secret = np.array([
    first_line, second_line, third_line, second_line, first_line
])
big_original = big_secret.copy()
first = [x ** 0.5 for x in range(100)]
second = [x ** (1 / 3) for x in range(100, 200)]
third = [x / y for x in range(200, 300, 2) for y in [3, 5]]
great_secret = np.array([first, second, third]).T
```


big_secret массивінің әр жолында 51 сан бар: x мәндерінің 17 элементтік тізбегі үш y мәнімен үйлеседі. Транспонирлегеннен кейін great_secret 100 нысан мен үш белгіге ие. Бұлар нақты өлшеулер емес, математикалық құрылымдар. Олардың бағандарына жас немесе құн мағынасын бермеңіз. np.array құруға дейін құрамдастар ұзындығын тексеру range оң жақ шегіндегі қатені табуға көмектеседі.



```python
students = np.array([
    [1, 135, 34, 4.0], [2, 160, 43, 5.0],
    [3, 163, 40, 4.3], [4, 147, 44, 5.0],
    [5, 138, 41, 4.7], [6, 149, 54, 3.9],
    [7, 136, 39, 4.2], [8, 154, 48, 4.9],
    [9, 137, 35, 3.7], [10, 165, 60, 4.6],
], dtype=float)
```


students бағандары: журналдағы оқу нөмірі, бойы см-мен, салмағы кг-мен, орташа балы. Нөмір — идентификатор, ол сандық түсіндіруші белгіге айналмауы тиіс. Кесте шағын және есептеуді жаттықтыруға арналған; он жолға сүйеніп барлық мектеп оқушысы туралы қорытынды жасауға болмайды.

## Міндетті тапсырмалар

1. **Массивтер паспорты.** big_secret, great_secret және students үшін shape, ndim, size және dtype шығарыңыз. Құру коды бойынша әр ось ұзындығының шығу тегін түсіндіріңіз. big_secret=(5, 51), great_secret=(100, 3), students=(10, 4) пішіндерін тексеріңіз. Бір скалярды, бір жолды және екінші осін сақтап бір бағанды алуды көрсетіңіз. Әр жағдайда алдымен пішінді болжап, кейін нақты пішінмен салыстырыңыз.

2. **Бастапқы матрицаның диагональдары.** big_original қолданып, соңғы баған қосындысын, алғашқы бес бағанның бас диагоналі қосындысын және соңғы бес бағанның бас диагоналі көбейтіндісін табыңыз. Бөліктер 5x5 пішінінде болуы тиіс. Тілімдер мен np.diag қолданыңыз; бір шама үшін қысқа циклмен тәуелсіз тексеру қосыңыз. Есепте -5: неге соңғы бес бағанды алатынын, ал :-5 басқа есепті шешетінін түсіндіріңіз. Көбейтіндіні қосындымен алмастырмаңыз және операцияға дейін элементтерді дөңгелектемеңіз.

3. **Индекстердің жұптығына қарай өзгерту.** big_original массивінің тәуелсіз көшірмесі ретінде big_changed құрыңыз. Екі индексі де жұп элементтерді -1, екі индексі де тақ элементтерді 1 мәнімен алмастырып, қалғандарын сақтаңыз. Жұптық жолдардың адамша нөміріне емес, нөлден басталатын индекстерге қатысты. 2-тапсырмадағы екі диагональдық шаманы қайта табыңыз. big_original өзгермегенін тексеріп, орындардың үш тобының әрқайсысына бір бақылау элементін келтіріңіз. Өзгерту үшін толық кірістірілген цикл қолданбаңыз.

4. **Маска, функция және жалпақ пішін.** great_secret үшін бірінші жол косинустарының қосындысын және 50-ден үлкен барлық мәндер қосындысын есептеңіз. Таңдалған элементтер санын табыңыз. Әдеттегі жол бойынша ретпен тәуелсіз жалпақ көшірме жасап, 150 индексіндегі мәнді анықтаңыз. Оның бастапқы матрицаның қай жолы мен бағанынан келгенін көрсетіңіз. Содан кейін әр бағанды бөлек сұрыптап, соңғы жол қосындысын табыңыз. Мұндай сұрыптау жолдарды тұтас нысандар ретінде неге сақтамайтынын түсіндіріңіз.

5. **Функция және reshape реті.** 1-25 сандарынан 5x5 пішінді my_array құрыңыз. Орталық 3x3 шаршыны алып, оның бірінші жолы [7, 8, 9] екенін растаңыз. Әр санның синусы ретінде my_sin құрып, қосындысын есептеңіз. `my_changed = my_sin.copy()` тәуелсіз көшірмесін жасаңыз. my_changed массивіндегі орталық 3x3 шаршыны бірліктермен алмастырып, жаңа қосындыны табыңыз. Дәл my_changed массивінің алғашқы төрт бағанын 10x2 массивіне айналдырып, бірінші баған қосындысын табыңыз. Пішіндерді және нәтиженің алғашқы екі жолының шығу тегін жазыңыз. Бастапқы my_sin сақталсын.

6. **Өлшеулер статистикасы.** students бойынша салмақ медианасын, орташа және медианалық салмақ айырмасын, салмақ дисперсиясын және орташа балдың стандартты ауытқуын табыңыз. var мен std үшін міндетті түрде ddof=0 көрсетіп, содан кейін кемінде бір шаманы ddof=1 арқылы көрсетіңіз. Ауытқулар квадраттары арқылы берілген формуламен салмақ дисперсиясын қайта есептеп, нәтижелерді салыстырыңыз. Журнал нөмірін шығарып, үш өлшенетін белгінің корреляция матрицасын құрыңыз. Әртүрлі белгілер арасындағы мәні бойынша ең кіші корреляцияны және бөлек модулі бойынша ең кішісін табыңыз; диагональ мен симметриялы қайталануларды есепке алмаңыз.

7. **Масштаб және қайта орындау мүмкіндігі.** Жұп шектерді қоспай, 100-ден 1000-ға дейінгі барлық тақ бүтін сандар квадраттарынан bigdata құрыңыз. Медиананы, ddof=0 берілген std мәнін, жұп және тақ индекстердегі элементтер корреляциясын есептеңіз. Неліктен квадраттардың өз жұптығы емес, индекстер салыстырылатынын түсіндіріңіз. Жеке тәжірибе үшін seed=42 болатын rng және 0-ден 9-ға дейінгі бүтін сандардың 4x3 матрицасын жасаңыз. Сол seed арқылы генераторды қайта құрып, сәйкестікті тексеріңіз; бар генератордың келесі шақырылуын тізбектің басы емес, жалғасы деп түсініңіз.

8. **Нақты белгілер матрицасы.** football кестесін төменде берілген үлгімен жүктеп, NumPy ортасына Acceleration, SprintSpeed, Stamina өрістерін осы ретпен беріңіз. Тәуелсіз float массивін құрып, пішіні мен ақырлылығын тексеріңіз. Бағандар бойынша орташа мәндер мен std табыңыз. Әр бағанды broadcasting арқылы орталықтандырыңыз; нәтиженің орташа мәні нөлге allclose арқылы тең екенін тексеріңіз. Әр жолды орталықтандыру қалай ерекшеленетінін және осы есепте (n, 3) пен (n,) пішіндері автоматты түрде неге үйлеспейтінін түсіндіріңіз.

9. **Индекс және түсіндіру.** Сол үш дағды үшін @ арқылы 0.4, 0.4, 0.2 салмақтары бар оқу индексін құрыңыз. argsort арқылы ең үлкен бес нәтиженің индекстерін тауып, оларды орын бойынша аттармен сәйкестендіріңіз. Атты, үш дағдыны және индексті top5 кестесінде сақтаңыз. Бұл — шартты салмақталған қосынды, ойыншының барлық жағынан жақсы екенінің дәлелі емес. Салмақтар мен тарихи кескін шектеулерін түсіндіріңіз; жолдарды ауыстыру ат пен сипаттамалар сәйкестігін бұзбағанын тексеріңіз.
