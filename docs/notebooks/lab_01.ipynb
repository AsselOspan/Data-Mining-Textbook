Нәтиже — орындалатын ноутбук немесе скрипт, деректер паспорты және нәтижелер CSV файлы. Орындау кезінде экраныңызды көрмеген адамға да нәтиже түсінікті болуы тиіс.

Жұмыс бастапқы зертханалық жұмыстағы Series құру, олармен амалдар орындау, DataFrame жасау және CSV алмасу тапсырмаларын дамытады. Күрделі топтау мен біріктіру келесі сабақтарға ауыстырылған: мұнда деректер пішіні мен индекстің мағынасын түсіну маңыздырақ. Нәтижені қайталау үшін кездейсоқ мысалдарға бекітілген генератор қолданылады.

## Құралдар, деректер және дайындық

Python, pandas, NumPy және код редакторы не Jupyter жеткілікті. Орындау кезінде интернет қажет емес. data/football.csv файлы жинаққа енгізілген. Онда 12897 жол, 42 баған бар және жетіспейтін ұяшықтар жоқ. Бұл футболшылар рейтингтерінің тарихи жиыны; ол командалардың қазіргі құрамдарын сипаттамайды.

Name — дереккөздегі аты, оның бірегей идентификатор екені дәлелденбеген. Age — жас. Position, Club, Nationality — санаттық белгілер. Unnamed: 0 — бұрын сақталған индекс. Wage пен Value тек дереккөз бірліктерінде талданады: валюта мен кезең метадеректермен расталмаған. Penalties — соғылған гол саны емес, пенальти орындау дағдысының рейтингі. Бұл шектеулерді код түсіндірмесіне ғана емес, өз паспортыңызға да енгізіңіз.

Кодты data каталогы орналасқан жинақ түбірінен іске қосыңыз. Дайындық блогын өз жұмысыңызда қолдануға болады:

## Талданған мысал

Үш шартты құрал өлшеулер санын жіберді. Series құрып, әр мәнді арттырып, деректерді индекс белгісі мен орны бойынша таңдаймыз. Мысал төмендегі тапсырмалардан өзгеше; ол дайын жауаптарды емес, пайымдау тәсілін көрсетеді.



```python
counts = pd.Series(
    [4, 7, 5], index=['sensor_c', 'sensor_a', 'sensor_b']
)
updated = counts + 2
print(updated.loc['sensor_a'])
print(updated.iloc[0])
table = pd.DataFrame({'before': counts, 'after': updated})
print(table.shape)
```


Нәтижелер: 9, 6, `(3, 2)`. Бірінші команда sensor_a атауын іздейді, екіншісі мұнда sensor_c сәйкес келетін бірінші орынды алады. DataFrame бір көрсеткіштің екі нұсқасын қамтиды. Оның индексі жазбаларды құрал атаулары бойынша біріктіріп тұр. Бір Series ретін кездейсоқ ауыстырсаңыз, сәйкестік белгілер бойынша сақталады.

Әрбір өз амалыңыз үшін алдымен нәтиженің типі мен пішінін болжаңыз, содан кейін кодты орындап, болжамды шығарылыммен салыстырыңыз. Мұндай қысқа түсіндірме түсіндірілмеген көптеген кестеден маңыздырақ.

## Тапсырмалар

### 1. Series нысанын үш тәсілмен құру

`a, b, c, d, e` белгілерімен `[2, 4, 6, 8, 10]` тізімінен `s_list`; `{'a': 10, 'b': 8, 'c': 6, 'd': 4, 'e': 2}` сөздігінен `s_dict`; сол белгілер үшін 3 санымен `s_constant` құрыңыз. Әр Series мәндерін, индексін, dtype және ұзындығын көрсетіңіз. Конструкторға қандай деректер берілетінін, ал қайсысы автоматты пайда болатынын түсіндіріңіз. Тексеру нәтижесі: үш нысанның да ұзындығы 5 және белгілер жиыны бірдей.

### 2. Қол жеткізу және сәйкестендіру

`s_list` ішінен c белгісі бар элементті және бірінші элементті алыңыз; әртүрлі айқын индексатор қолданыңыз. `s_list` пен `s_dict` қосыңыз. `s_dict` нұсқасын e, d, c, b, a ретімен құрып, қосуды қайталаңыз. Нәтижелерді белгілер бойынша салыстырып, сәйкестігін түсіндіріңіз. Содан кейін Series нысандарының біріне f белгісін қосып, сәйкес мән болмағанда не болатынын көрсетіңіз. Пайда болған жетіспейтін мәннің себебін түсіндірмей тұрып оны толтырмаңыз.

### 3. Өзіңіздің жазбалар кестеңіз

`student_id`, `Name`, `Age`, `Hours` өрістері бар сөздіктен `students` DataFrame құрыңыз. Төрт жазба, бірегей бүтін student_id, ойдан шығарылған аттар және теріс емес Hours пайдаланыңыз. Мазмұнын өзіңіз таңдап, есепте келтіріңіз. `(4, 4)` пішінін, бағандар ретін және Age типін тексеріңіз. Неліктен student_id орташа мәнін есептемеу керектігін және Name идентификаторды неге алмастырмайтынын түсіндіріңіз.

### 4. Бір және екі өлшем

students ішінен Age бағанын алдымен Series, кейін DataFrame ретінде алыңыз. Әр нәтиже үшін `type`, `shape`, `ndim` жазыңыз. `.iloc` арқылы бір жол таңдап, онда әртүрлі белгілердің мәндері біріктірілгенін көрсетіңіз. Бастапқы бағандардың типтері туралы тек осы жолдың dtype мәніне сүйеніп қорытынды жасамаңыз: аралас кестенің жолы ортақ көрсету типіне ие болуы мүмкін.

### 5. Қайталанатын кездейсоқ кесте

A, B, C, D, E бағандары бар `random_table` құру үшін `rng.random((5, 5))` қолданыңыз. Үшінші жолды, C бағанын және A мәні 0.5-тен үлкен жолдарды көрсетіңіз. Сүзгілеуге `.loc` қолданыңыз. Бағандар қосындысын айқын `axis=0`, кейін жолдар қосындысын `axis=1` арқылы есептеңіз. Екі нәтиженің қосындыларын массивтің жалпы қосындысымен `np.isclose` арқылы салыстырыңыз. Сүзгілеуден кейінгі таңдама көлемін болжау талап етілмейді: оны есептеп, неліктен 0-ден 5-ке дейін бола алатынын түсіндіріңіз.

### 6. Жинақтағы CSV файлын тексеру

football ішінен алғашқы үш және соңғы екі жазбаны алыңыз. `shape`, `columns`, `dtypes` шығарыңыз; `info()` бөлек шақырыңыз. Name, Age, Club, Position, Penalties өрістерінің барын және жетіспейтін мәндердің жоғын тексеріңіз. Пішіннің `(12897, 42)` екеніне көз жеткізіңіз. Паспортқа сандық, санаттық және техникалық белгінің әрқайсысынан бір мысал енгізіңіз. SprintSpeed мәнін км/сағ, ал Penalties мәнін соғылған голдар деп неліктен түсіндіруге болмайтынын бөлек жазыңыз.

### 7. Шағын аналитикалық карточка

Age минимумын, максимумын, орташа мәнін, медианасын және бос емес мәндер санын есептеңіз. `n_rows`, `age_min`, `age_max`, `age_mean`, `age_median` өрістері бар бір жолдық `age_summary` DataFrame құрыңыз. Орташа мән мен медиана минимум мен максимум арасында екенін тексеріңіз. Бастапқы бағанды емес, тек көрсетілетін нәтижені дөңгелектеңіз. Екі сөйлем жазыңыз: карточка нені көрсетеді және барлық футболшылардың бас жиынтығы туралы нені айтпайды.

### 8. Сақтау және қайта оқу

students кестесін `index=False` параметрімен `work_results/lab01_students.csv` файлына сақтаңыз, қайта оқып, жол санын, баған атауларын және мазмұнын салыстырыңыз. age_summary кестесін `work_results/lab01_age_summary.csv` файлына сақтаңыз. Индексті әдепкі параметрмен экспорттағанда қандай артық өріс пайда болуы мүмкін екенін түсіндіріңіз. Соңында бүкіл ноутбукты таза ядродан қайта орындаңыз не скриптті қайта іске қосыңыз: тексеру тек файлдағы кодқа сүйенуі тиіс.

## Не тапсыру және қалай тексеру керек?

`lab01.ipynb` не `lab01.py`, деректер паспорты бар қысқа есеп және екі CSV ұсыныңыз. Кодта импорттар, Path арқылы жүктеу, нәтижелер каталогын құру және дербес тексерулер болуы тиіс. Есепте әр тапсырманың шағын шығарылымын көрсетіңіз; барлық 12897 жолды кірістірмеңіз.

| Файл | Деректер жолдары | Бағандар | Негізгі тексеру |
|---|---|---|---|
| lab01_students.csv | 4 | student_id, Name, Age, Hours | student_id бірегей |
| lab01_age_summary.csv | 1 | n_rows, age_min, age_max, age_mean, age_median | n_rows football жолдарының санына тең |

CSV құрамында қызметтік индекс болмауы тиіс. Жылжымалы нүктелі сандарды салыстыруға `np.isclose` қолдануға болады, өйткені мәтіндік түрде жазып, қайта оқығанда өте шағын айырмалар болуы мүмкін. assert болуы түсіндірмені алмастырмайды: тексерумен бірге оның қандай қасиетті растайтыны жазылуы тиіс.

Қорытындыда үш нәтижені тұжырымдаңыз: Series тізімнен қалай ерекшеленеді, DataFrame жазбаларды қалай ұйымдастырады, деректердің қандай шектеулерін ескеру қажет. Нәтиже тексеруден өтпесе, қате хабарын, себебін және түзетуін келтіріңіз. Тексеруді жойып, қатені жасыруға болмайды.

## Жиі кездесетін қиындықтар және қосымша тапсырма

KeyError пайда болса, таңдалған белгіні `index` не `columns` мәндерімен салыстырыңыз. Баған саны бірге тең болса, оқу параметрлері мен файлды тексеріңіз. Кесте сипаттамасынан кейін None шықса, сіз `info()` қайтаратын мәнді басып шығарып отырған болуыңыз мүмкін. Жолдар мен бағандар қосындысы аздап өзгеше болса, деректерді алмастырмай, жуық салыстыру қолданыңыз.

Қосымша тапсырма. Үш белгінің тек екеуі ғана сәйкес келетін екі Series нысанынан DataFrame құрыңыз. Жол санын және жетіспейтін мәндердің орнын болжап, болжамды тексеріңіз. Мұнда сәйкестендіру орын бойынша шартсыз жалғаудан неліктен пайдалырақ екенін түсіндіріңіз. Бұл тапсырма міндетті жаңа құрал ретінде емес, жалпы шкала аясындағы талдауды тереңдету ретінде бағаланады.
