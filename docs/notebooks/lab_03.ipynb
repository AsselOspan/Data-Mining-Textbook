
Әр кестенің жолы нені білдіретінін көрсетіңіз. groupby кейін ол жеке ойыншы емес, топ; value_counts кейін санат болады. Кездейсоқ индекстерді нәтижеге көшірмеңіз және автоматты пайда болған бағандарды тексермей, нөмірі бойынша қайта атамаңыз. Өріс атауларын айқын беріңіз.

## Талданған мысал: ішкі таңдамадағы үлес



```python
demo = pd.DataFrame({
    'Team': ['R', 'R', 'S', 'S', 'S'],
    'Role': ['A', 'B', 'A', 'A', 'B']
})
part = demo.loc[demo['Team'].eq('S')]
counts = part['Role'].value_counts()
shares = part['Role'].value_counts(normalize=True)
print(counts.to_dict())
print(shares.to_dict())
```


S командасы үшін A жиілігі 2, B жиілігі 1. Үлестер 2/3 және 1/3-ке тең. Бөлім — толық demo ішіндегі бес жол емес, S командасының үш жолы. «Барлық жазбаның қандай пайызы S және A санаттарына жатады?» деген сұрақта бөлім беске тең болады. Үлесті есептемей тұрып алымы мен бөлімін сөзбен жазыңыз.

Қорытындыны нақты құрам көлемі емес, жиында қамтылу ретінде тұжырымдаңыз.

Әр іріктелген позицияға жол саны мен пайызы көрсетілген сөйлем жазыңыз. share бағаны үшін 10 шегін қолданбаңыз.

Age > 35 іріктеңіз. Club жиіліктерін анықтап, ең көп қамтылған барлық клубты және алынған жас таңдамасының көлемін табыңыз. Көшбасшы клубтар үшін үлкен жастағы жазбалар үлесін сол клубтың барлық жазбалары ішінде есептеңіз. Абсолют сан және үлес бойынша екі реттеуді салыстырыңыз. Әр үлестің бөлімін көрсетіңіз. Бір үлкен жастағы ойыншысы бар шағын клубтың үлесі жоғары болуы мүмкін, сондықтан жанында екі жазба саны да қажет.

### 7. Spain үшін 21 жастың үлесі

Nationality == 'Spain' іріктеңіз. Осы ішкі таңдамадағы Age == 21 жазбаларының пайызын екі тәсілмен есептеңіз: normalize=True жиіліктері арқылы және сәйкес жол санының ішкі таңдама көлеміне қатынасы арқылы.
